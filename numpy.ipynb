```python
#dtype of array
import numpy as np 
arr = np.array([1, 2, 3]) 
print(arr.dtype)  
```

```python
#creating float array
arr = np.array([1.5, 2.3, 3.7], dtype=np.float32) 
print(arr, arr.dtype) 
```

```python
#complex data type
arr = np.array([2 + 3j, 4 - 5j], dtype=np.complex64)  
print(arr, arr.dtype)                       #O
```

```python
#boolean data type
arr = np.array([True, False, True, False], dtype=np.bool_)
print(arr, arr.dtype)
```

```python
#string data type
import numpy as np
arr = np.array(['apple', 'banana', 'cherry'], dtype=str)
print(arr)
print(arr.dtype)
```

```python
#Object data type
arr = np.array([10,"hello",3.14], dtype=np.object_)
print(arr, arr.dtype)
```

```python
#date time data type
arr = np.array(["2024-03-14","2025-01-01"], dtype=np.datetime64)
print(arr, arr.dtype)
```

```python
#changing data type
arr = np.array([1, 2, 3])
new_arr = arr.astype(float)  
print(new_arr, new_arr.dtype)
```

```python
#converting string to int
arr = np.array(["10", "20", "30"]) 
new_arr= arr.astype(int) 
print(new_arr, new_arr.dtype)  
```
