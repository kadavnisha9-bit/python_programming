```python

class Student:
    def __init__(self):
        self.__name = ""
        self.__age = 0
    def set_name(self, name):
        self.__name = name
    def set_age(self, age):
        self.__age = age
    def get_name(self):
        return self.__name
    def get_age(self):
        return self.__age

s = Student()
s.set_name("Rahul")
s.set_age(20)

print("name:", s.get_name())
print("Age:", s.get_age())
```

```python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.__account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
            self.__balance += amount
            print("Deposited:", amount)

    def withdraw(self, amount):
            if amount <= self.__balance:
                self.__balance -= amount
                print("withdrawn:", amount)
            else:
                print("insufficient balance")

    def get_balance(self):
                return self.__balance

b = BankAccount('123456789', 5000)
b.deposit(1000)
b.withdraw(2000)
print("Balance:", b.get_balance())
```

```python
#single inheritance
class Animal:
    def sound(self):
        print("Animal makes a sound")
class Dog(Animal):
    def bark(self):
        print("Dog barks")

d=Dog()
d.sound()
d.bark()
```

```python
#single inheritance
class Parent:
    def parent(self):
        print("This is parent class")
class Child(Parent):
    def son(self):
        print("This is child class")
p=Parent()
p.parent()
p.son()
```

```python
#multiple inheritance
class Father:
    def father(self):
        print("This is father class")
class Mother:
    def mother(self):
        print("This is mother class")
class Child(Father, Mother):
    def child(self):
        print("This is child class")
c=Child()
c.father()
c.mother()
```

```python
#multilevel inheritance
class Grandparent:
    def grandparent(self):
        print("This is grandparent class")
class Parent(Grandparent):
    def parent(self):
        print("This is parent class")
class Child(Parent):
    def child(self):
        print("This is child class")
c=Child()
c.grandparent()
c.parent()
c.child()
```

```python
#multilevel inheritance
class India:
    def country(self):
        print("This is india")
class Maharashtra(India):
    def state(self):
        print("This is maharashtara")
class Satara(Maharashtra):
    def district(self):
        print("This is satara")
s=Satara()
s.country()
s.state()
s.district()
```

```python
class Parent:
    def parent(self):
        print("This is the parent class")

class Son(Parent):
    def child(self):
        print("This is first child")

class Daughter(Parent):
    def child(self):
        print("This is second child")

d = Daughter()
s = Son()

s.parent()
d.parent()
```

```python
#hybrid inheritance
class A:
    def method_A(self):
        print("This is a")
class B(A):
    def method_B(self):
        print("This is b")
class C(A):
    def method_C(self):
        print("This is c")
class D(B, C):
    def method_D(self):
        print("This is d")
obj=D()
obj.method_A()
```
