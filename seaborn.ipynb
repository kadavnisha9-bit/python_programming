```python
#Scatter plot
import seaborn as sns  
import matplotlib.pyplot as plt 
# Sample Data 
tips = sns.load_dataset("tips") 
# Scatter Plot 
sns.scatterplot(x="total_bill", y="tip", hue="day", style="sex", size="size", data=tips)
plt.show() 
```

```python
#Line plot
import seaborn as sns
import matplotlib.pyplot as plt 
# Sample data 
fmri = sns.load_dataset("fmri")
# Create line plot 
sns.lineplot(x="timepoint", y="signal", hue="event", style="region", data=fmri)
# Show plot 
plt.show() 
```

```python
#Bar plot
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create bar plot 
sns.barplot(x="day", y="total_bill", hue="sex", data=tips) 
# Show plot 
plt.show() 
```

```python
#violin plot
import seaborn as sns
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create violin plot 
sns.violinplot(x="day", y="total_bill", hue="sex", data=tips, split=True) 
# Show plot 
plt.show()
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt
# Load dataset 
tips = sns.load_dataset("tips")
# Create strip 
sns.stripplot(x="day", y="total_bill", hue="sex", data=tips, jitter=True, dodge=True) 
# Show plot 
plt.show() 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create swarm plot 
sns.swarmplot(x="day", y="total_bill", hue="sex", data=tips, dodge=True) 
# Show plot 
plt.show() 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create point plot 
sns.pointplot(x="day", y="total_bill", hue="sex", data=tips, dodge=True, markers=["o", "s"], linestyles=["-", "--"]) 
# Show plot 
plt.show() 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create histogram 
sns.histplot(tips["total_bill"], bins=20, kde=True, color="blue") 
# Show plot 
plt.show() 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create KDE plot 
sns.kdeplot(x=tips["total_bill"], fill=True, color="blue", linewidth=2) 
# Show plot 
plt.show() 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips") 
# Create pair plot 
sns.pairplot(tips, hue="sex", diag_kind="kde") 
# Show plot 
plt.show() 
```

```python
import seaborn as sns 
import matplotlib.pyplot as plt 
# Load dataset
tips = sns.load_dataset("tips") 
# Select only numeric columns for correlation matrix 
corr = tips.select_dtypes(include=["number"]).corr() 
# Create heatmap 
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5, fmt=".2f")
# Show plot 
plt.show() 
```

```python
import seaborn as sns
import matplotlib.pyplot as plt 
# Load dataset 
tips = sns.load_dataset("tips")
# Create FacetGrid
g = sns.FacetGrid(tips, col="day", hue="sex", col_wrap=2)
# Map KDE plot to the grid
 g.map_dataframe(sns.kdeplot, x="total_bill", fill=True, alpha=0.5) 
# Add legend 
g.add_legend() 
# Show plot
plt.show() 
```
