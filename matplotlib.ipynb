```python
#Loading data
import pandas as pd 
 # Define column names  
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']
# Load dataset from URL
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
df = pd.read_csv(url, names=columns)
# Show first 5 rows
print(df.head())
```

```python
#Line plot
import matplotlib.pyplot as plt
plt.plot(df.index, df["sepal_length"], label="Sepal Length", color="b")
plt.plot(df.index, df["sepal_width"], label="Sepal Width", color="r")
plt.xlabel("Index")
plt.ylabel("Length (cm)")
plt.title("Line Plot of Sepal Features")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Iris dataset
df = pd.read_csv("Iris (1).csv")

# Check the data
print(df.head())
print(df.columns)
```

```python
#Bar chart
df_grouped = df.groupby("Species")["SepalLengthCm"].mean()
plt.bar(df_grouped.index, df_grouped.values, color=["blue", "orange",
"green"])
plt.xlabel("Species")
plt.ylabel("Average Sepal Length (cm)")
plt.title("Bar Chart of Average Sepal Length per Species")
plt.show()
```

```python
#Histogram
plt.hist(df["SepalLengthCm"], bins=10, color="purple", alpha=0.7,
edgecolor="black")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.title("Histogram of Sepal Length")
plt.show()
```

```python
#Scatter plot
plt.scatter(df["SepalLengthCm"], df["SepalWidthCm"], color="red", alpha=0.7)
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Scatter Plot of Sepal Length vs Width")
plt.show()
```

```python
#Pie chart
df_species_count = df["Species"].value_counts()
plt.pie(df_species_count, labels=df_species_count.index,
autopct="%1.1f%%", colors=["blue", "orange", "green"])
plt.title("Pie Chart of Species Distribution")
plt.show()
```

```python
#Box plot
plt.boxplot(df["SepalLengthCm"])
plt.xlabel("Sepal Length")
plt.ylabel("Values")
plt.title("Box Plot of Sepal Length")
plt.show()
```

```python
#customizing a plot(line plot)
# Customize Line Plot
plt.figure(figsize=(8,5)) # Set figure size
plt.plot(df.index, df["SepalLengthCm"], label="Sepal Length", color="blue",
linestyle="--", marker="o", markersize=5)
plt.plot(df.index, df["SepalWidthCm"], label="Sepal Width", color="red",
linestyle="-.", marker="s", markersize=5)
# Add labels, title, legend, grid
plt.xlabel("Index (Observations)", fontsize=12, fontweight='bold')
plt.ylabel("Length (cm)", fontsize=12, fontweight='bold')
plt.title("Customized Line Plot of Sepal Features", fontsize=14,
fontweight='bold')
plt.legend()
plt.grid(True, linestyle="--", alpha=0.5)
# Show plot
plt.show()
```

```python
#Subplots
fig, axes = plt.subplots(2, 2, figsize=(10,8)) # 2x2 grid of subplots
# Histogram of Sepal Length
axes[0, 0].hist(df["SepalLengthCm"], bins=10, color="purple", alpha=0.7,
edgecolor="black")
axes[0, 0].set_title("Histogram of Sepal Length")
# Scatter Plot of Sepal Length vs Sepal Width
axes[0, 1].scatter(df["SepalLengthCm"], df["SepalWidthCm"], color="red", alpha=0.7)
axes[0, 1].set_xlabel("SepalLengthCm")
axes[0, 1].set_ylabel("SepalWidthCm")
axes[0, 1].set_title("Scatter Plot of Sepal Length vs Width")
# Box Plot of Petal Length
axes[1, 0].boxplot(df["PetalLengthCm"])
axes[1, 0].set_title("Box Plot of Petal Length")
# Bar Chart of Average Sepal Length per Species
df_grouped = df.groupby("Species")["SepalLengthCm"].mean()
axes[1, 1].bar(df_grouped.index, df_grouped.values, color=["blue", "orange",
"green"])
axes[1, 1].set_xlabel("Species")
axes[1, 1].set_ylabel("Avg Sepal Length")
axes[1, 1].set_title("Bar Chart of Sepal Length per Species")
plt.tight_layout() # Adjust layout
plt.show()
```
