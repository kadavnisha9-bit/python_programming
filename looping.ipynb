```python
#count digit of no using while loop
num=123456
count = 0
while num > 0:
    count +=1
    num//=10
print("digits:",count)
```

```python
#write a while to find the sum of digits of a number
num = 123
sum_digits = 0
while num >0:
    sum_digits += num % 10
    num //=10
print(sum_digits)
```

```python
#print sum of element in a list using for loop
numbers = [1,2,3,4,5]
sum_numbers=0
for num in numbers:
    sum_numbers += num
print(sum_numbers)
```

```python
#print all element of a list except the number 3 using for loop
numbers = [1,2,3,4,5]
for num in numbers:
    if num ==3:
        continue
    print(num)
```
