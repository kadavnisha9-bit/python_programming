```python
# Create dictionary with key-value pairs
mydict = {"name": "Nisha", "age": 25, "city": "New York"}
print("Keys:", mydict.keys())
print("values:",mydict.values())
```

```python
#merge two dictionaries
dict1={"a":1,"b":2}
dict2={"c":3,"d":4}
merged_dict={**dict1,**dict2}
print("merged dictonary:",merged_dict)
```

```python
# Count frequency of each word
text = "apple banana apple orange banana apple"

words = text.split()
word_count = {}

for word in words:
    word_count[word] = word_count.get(word, 0) + 1

print("Word frequency:", word_count)
```

```python
# Find the key with the highest value in a dictionary
scores = {"alice": 85, "bob": 92, "charlo": 99}

max_key = max(scores, key=scores.get)

print("Key with max value:", max_key)
```

```python
# Check whether the given key exists
student = {"name": "john", "grade": "A", "age": "20"}

key_to_check = "grade"

if key_to_check in student:
    print(f"'{key_to_check}' exists in the dictionary.")
else:
    print(f"'{key_to_check}' does not exist in the dictionary.")
```

```python
# Sort dictionary in ascending order by values
marks = {"math": 78, "english": 85, "science": 72}

sorted_dict = dict(sorted(marks.items(), key=lambda item: item[1]))

print("Sorted dictionary:", sorted_dict)
```

```python
# Sum of all values
expenses = {"rent": 1200, "groceries": 300, "internet": 50}

total_expense = sum(expenses.values())

print("Total expenses:", total_expense)
```

```python
#swap keys and values
dict1 ={"a":1, "b":2, "c":3}
dict2 = {"b":5, "c":7, "d":9}
common_keys = dict1.keys()&dict2.keys()
print("common keys:", common_keys)
```
