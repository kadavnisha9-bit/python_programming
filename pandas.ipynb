```python
import pandas as pd  
# Define column names 
columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species'] 
 # Load dataset from URL 
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data' 
df = pd.read_csv(url, names=columns)  
# Show first 5 rows 
print(df.head()) 
```

```python
# Show first 5 rows 
print(df.head()) 
# # Show summary statistics 
print(df.describe())
#  # Show data types 
print(df.dtypes) 
```

```python
setosa_df = df[df['species'] == 'Iris-setosa'][['sepal_length', 'sepal_width']] 
print(setosa_df.head()) 
```

```python
import numpy as np 
# Check for missing values 
print("Missing values before:", df.isnull().sum())
# Fill missing values with mean 
df.fillna(df.mean(numeric_only=True), inplace=True)
# Check again 
print("Missing values after:", df.isnull().sum()) 
```

```python
df['sepal_area'] = df['sepal_length'] * df['sepal_width'] 
print(df.head()) 
```

```python
grouped_df = df.groupby('species').mean() 
print(grouped_df) 
```

```python
extra_data = pd.DataFrame({   
    'species': ['Iris-setosa', 'Iris-versicolor', 'Iris-virginica'],  
    'color': ['red', 'green', 'blue'] }) 
 # Merge both DataFrames 
merged_df = pd.merge(df, extra_data, on='species')
print(merged_df.head()) 
```
