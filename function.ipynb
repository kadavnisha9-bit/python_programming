```python
#function argument
#function defination with a parameter
def greet(name):
    print("hello,"+ name)
    #function call with an argument
    greet("nisha")
```

```python
#calculate the area of circle
import math 
def circle_area(radius):
    return math.pi * radius **2
print(circle_area(5))
```

```python
#count the number of words in a sentence
def count_words(sentence):
    return len(sentence.split())
print(count_words("Hello nisha, welcome to python"))
```

```python
#find the largest even no
def largest_even(numbers):
    evens = [num for num in numbers if num % 2 ==0]
    return max(evens)
print(largest_even([3,7,2,8,10,15]))
```

```python
#find the factorial using lambda
from functools import reduce
factorial = lambda n: reduce(lambda x,y: x*y, range(1, n+1))if n > 0 else 1
print(factorial(5))
```

```python
#convert a list of strings to uppercase using map()
def to_uppercase(words):
    return list(map(str.upper,words))
print(to_uppercase(["hello", "world"]))
```

```python
#filter odd numbers from a list
def filter_odds(numbers):
    return list(filter(lambda x: x % 2 != 0, numbers))
print(filter_odds([1,2,3,4,5]))
```

```python
#sum of all numbers using reduce
from functools import reduce
def sum_numbers(numbers):
    return reduce (lambda x, y: x + y, numbers)
print(sum_numbers([1,2,3,4,5]))
```

```python
#find palindrome words in a list
def filter_palidromes(words):
    return list(filter(lambda word: word == word[::-1], words))
print(filter_palidromes(["madam", "hello", "racecar", "python"]))
```

```python
#merge two list using map
def merge_lists(list1, list2):
    return list(map(lambda x,y: x+y, list1, list2))
print(merge_lists([1,2,3],[4,5,6]))
```

```python
#find the length of each word in a list
def word_lengths(words):
    return list(map(len, words))
print(word_lengths(["helllo", "world"]))
```

```python
#find the common elements in two list
def common_elements(list1, list2):
    return list(filter(lambda x: x in list2, list1))
print(common_elements([1,2,3],[2,3,4]))
```

```python
#find the square of even numbers using map()
def even_squares(numbers):
    return list(map(lambda x: x ** 2, filter(lambda x: x % 2 == 0, numbers)))
print(even_squares([1,2,3,4]))
```

```python
#check if all elements in a list are even
def all_even(numbers):
    return all(map(lambda x: x % 2 ==0, numbers))
print(all_even([2,4,6]))
print(all_even([2,3,6]))
```

```python
def find_primes_in_range(start, end):
    primes = []
    for  num in range(start, end + 1):
          count = 0
          for i in range(1, num + 1):
               if num % i ==0:
                    count +=1
          if count == 2:
               primes.append(num)
    return primes
print(find_primes_in_range(10, 50))
```

```python
def longest_word(sentence):
    words = sentence.split()
    longest = words[0]
    for word in words:
        if word in words:
            if len(word) > len(longest):
                longest = word
    return longest
print(longest_word("I love learning python programming"))
```

```python
def is_sorted(numbers):
    if numbers == sorted(numbers):
        return True
    else:
        return False

print(is_sorted([1,2,3,4,5]))
print(is_sorted([5,3,2,1]))
```

```python
def multiplication_table(n):
    for i in range(1, n + 1):
       print("Table of", i)
       for j in range(1, n +1 ):
           print(i, "x", j, "=", i * j)
    print()
multiplication_table(5)
```

```python
def count_frequency(list1):
    d = {}
    for i in list1:
        d[i] = list1.count(i)
    return d
list1 = ["apple", "banana", "apple", "orange", "banana", "apple"]
print(count_frequency(list1))
```

```python
def sum_of_digits_list(numbers):
    result = []
    for num in numbers:
        total = 0
        for digit in str(num):
            total = total + int(digit)
        result.append(total)
print(sum_of_digits_list([34, 12,45]))
```

```python
def find_palidrome_numbers(start, end):
    result = []
    for i in range(start, end + 1):
        if str(i) == str(i)[::-1]:
            result.append(i)
    return result
print(find_palidrome_numbers(100, 200))
```
